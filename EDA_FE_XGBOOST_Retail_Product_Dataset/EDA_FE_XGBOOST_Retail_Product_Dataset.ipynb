```python
import pandas as pd
import plotly.express as px
```

```python
url = "https://raw.githubusercontent.com/kamyczuram/Projects/main/Data/synthetic_dataset.csv"
df = pd.read_csv(url)
```

```python
fig = px.histogram(df,
                   x="Price",
                   color="Category",
                   facet_col="Category",
                   marginal="box",
                   nbins=30,
                   opacity=0.75)

fig.update_layout(height=600, width=1000)
fig.show()
```

```python
fig = px.histogram(df,
                   x="Price",
                   color="Stock",
                   facet_col="Stock",
                   marginal="box",
                   nbins=30,
                   opacity=0.75)

fig.update_layout(height=600, width=1000)
fig.show()
```

```python
numeric_df = df.select_dtypes(include='number')
correlation_matrix = numeric_df.corr()
print(correlation_matrix)
```

```python
categorical_cols = ['Stock', 'Category']
numeric_cols = df.select_dtypes(include='number').columns

for cat_col in categorical_cols:
    print(f"\n Grouped by: {cat_col}")
    grouped = df.groupby(cat_col)[numeric_cols].agg(['mean', 'median'])
    print(grouped)
```

```python
missing_mask = df.isnull().astype(int)
```

```python
num_df = df.select_dtypes(include='number')
```

```python
combined_df = pd.concat([missing_mask, num_df], axis=1)
```

```python
combined_df.corr()
```
