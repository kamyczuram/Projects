```python
import pandas as pd
import plotly.express as px
```

```python
url = "https://raw.githubusercontent.com/kamyczuram/Projects/main/Data/synthetic_dataset.csv"
df = pd.read_csv(url)
```

```python
fig = px.histogram(df,
                   x="Price",
                   color="Category",
                   facet_col="Category",
                   marginal="box",
                   nbins=30,
                   opacity=0.75)

fig.update_layout(height=600, width=1000)
fig.show()
```
